```python
class0=iris[iris['target']==0]
class1=iris[iris['target']==1]
class2=iris[iris['target']==2]
categories=['sepal length (cm)','sepal width (cm)','petal length (cm)','petal width (cm)']
n_bins=40
```

```python
for l in categories :

    x=[class0[l],class1[l],class2[l]]
    fig, ax0 = plt.subplots(nrows=1)
    colors = ['blue', 'orange', 'green']
    ax0.hist(x, n_bins, histtype='bar', color=colors, label=['class 0', 'class 1', 'class 2'])
    ax0.legend(prop={'size': 10})
    ax0.set_title('distribution of ' + l+' in diffrent classes')
    plt.ylabel('amount')
    plt.xlabel(l)
```
